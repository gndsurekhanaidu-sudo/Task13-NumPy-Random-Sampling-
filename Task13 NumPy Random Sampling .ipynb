```python
import numpy as np

rng = np.random.default_rng(42)

numbers = rng.integers(1, 11, size=5)

print(numbers)
```

```python
floats = rng.random(5)

print(floats)
```

```python
arr = np.array([10, 20, 30, 40, 50])

sample = rng.choice(arr, size=3, replace=False)

print(sample)
```
